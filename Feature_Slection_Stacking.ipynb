## Load the dataset



```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Data mining & ML/hypertension_dataset (1).csv")
display(df.head())
```

## Preprocess the data


```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np
import pandas as pd


df.isnull().sum()
```

```python
df
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd

# Create an imputer with a 'most_frequent' strategy
imputer = SimpleImputer(strategy='most_frequent')

# Reshape the 'Medication' column to be 2D as required by the imputer and flatten the output
df['Medication'] = imputer.fit_transform(df[['Medication']]).flatten()

# Check if there are any remaining missing values in the 'Medication' column
print("Missing values in 'Medication' after imputation:", df['Medication'].isnull().sum())
```

```python
X = df.drop('Has_Hypertension', axis=1)
y = df['Has_Hypertension']
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Identify categorical columns
categorical_features = X.select_dtypes(include=['object', 'category']).columns

one_hot_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Apply one-hot encoding to the categorical features
encoded_features = one_hot_encoder.fit_transform(X[categorical_features])

# Create a DataFrame from the encoded features with appropriate column names
encoded_feature_names = one_hot_encoder.get_feature_names_out(categorical_features)
df_encoded = pd.DataFrame(encoded_features, columns=encoded_feature_names, index=X.index)

# Drop the original categorical columns from the dataframe
X = X.drop(categorical_features, axis=1)

# Concatenate the original dataframe (without categorical columns) with the encoded features
df_processed = pd.concat([X, df_encoded], axis=1)

display(df_processed.head())
```

```python
from sklearn.preprocessing import StandardScaler

numerical_features_processed = df_processed.select_dtypes(include=['float64', 'int64']).columns.tolist()


# Initialize StandardScaler
scaler = StandardScaler()

# Apply scaling to the numerical features
df_processed[numerical_features_processed] = scaler.fit_transform(df_processed[numerical_features_processed])

display(df_processed.head())
```

```python
df_processed.shape
```

## Apply feature engineering


```python
# Create interaction features
df_processed['Salt_Stress_Interaction'] = df_processed['Salt_Intake'] * df_processed['Stress_Score']
df_processed['Age_BMI_Interaction'] = df_processed['Age'] * df_processed['BMI']

# Display the head of the DataFrame with the new features
display(df_processed.head())
```

## Apply feature selection


```python
from sklearn.feature_selection import SelectKBest, f_classif
import matplotlib.pyplot as plt
import pandas as pd


# Use SelectKBest with f_classif to calculate scores for all features
selector = SelectKBest(score_func=f_classif, k='all')
selector.fit(df_processed, y)

# Get the scores
feature_scores = selector.scores_

# Create a pandas Series for better visualization and sorting
feature_scores_series = pd.Series(feature_scores, index=df_processed.columns)

# Print the scores
print("Feature scores (f_classif):")
print(feature_scores_series.sort_values(ascending=False))
```

```python
# Use SelectKBest with f_classif to select the top K features (keeping the original selection logic)
k = 15  # Choose a reasonable value for K
selector_kbest = SelectKBest(score_func=f_classif, k=k)
X_selected = selector_kbest.fit_transform(df_processed, y)

# Get the names of the selected features
selected_feature_indices = selector_kbest.get_support(indices=True)
selected_feature_names = df_processed.columns[selected_feature_indices]

# Convert the selected features back to a DataFrame with column names
X_selected_df = pd.DataFrame(X_selected, columns=selected_feature_names, index=df_processed.index)

display(X_selected_df.head())
print(f"\nSelected features ({k} best): {list(selected_feature_names)}")
```

## Split data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_selected_df, y, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

## Apply grid search for hyper-parameter tuning


```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC # Import SVC for SVM

# Define models and their hyperparameter grids
models = [
    ('LogisticRegression', LogisticRegression(solver='liblinear'), {'C': [0.1, 1.0, 10.0]}),
    ('RandomForestClassifier', RandomForestClassifier(random_state=42), {'n_estimators': [100, 200], 'max_depth': [None, 10, 20]}),
    ('GradientBoostingClassifier', GradientBoostingClassifier(random_state=42), {'n_estimators': [100, 200], 'learning_rate': [0.01, 0.1, 0.5]}),
    ('SVC', SVC(probability=True, random_state=42), {'C': [0.1, 1.0, 10.0], 'kernel': ['linear', 'rbf']}) # Added SVM with hyperparameter grid
]

best_estimators = {}

for name, model, param_grid in models:
    print(f"Running Grid Search for {name}...")
    grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
    grid_search.fit(X_train, y_train)

    print(f"Best parameters for {name}: {grid_search.best_params_}")
    print(f"Best cross-validation accuracy for {name}: {grid_search.best_score_:.4f}")

    best_estimators[name] = grid_search.best_estimator_
    print("-" * 30)

print("Grid search completed for all models.")
```

```python
for name, model in best_estimators.items():
    print(f"Model {name} is trained and ready for evaluation.")
```

## Apply stacking method


```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression


estimators = [
    ('RandomForestClassifier', best_estimators['RandomForestClassifier']),
    ('SVC', best_estimators['SVC']),
    ('GradientBoostingClassifier', best_estimators['GradientBoostingClassifier'])
]

# Define the final estimator (meta-model)
final_estimator = LogisticRegression(solver='liblinear')

# Instantiate the StackingClassifier
stacking_model = StackingClassifier(estimators=estimators, final_estimator=final_estimator, cv=5)

# Fit the StackingClassifier to the training data
stacking_model.fit(X_train, y_train)

print("Stacking model trained successfully.")
```

## Evaluate models


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

results = {}

# Evaluate individual models
for name, model in best_estimators.items():
    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_pred_proba = model.predict_proba(X_test)[:, 1]
    else:
        y_pred_proba = None # Or some other handling if probabilities are not available


    results[name] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, pos_label='Yes'),
        'Recall': recall_score(y_test, y_pred, pos_label='Yes'),
        'F1-score': f1_score(y_test, y_pred, pos_label='Yes'),
        'ROC AUC': roc_auc_score(y_test, y_pred_proba) if y_pred_proba is not None else 'N/A'
    }

# Evaluate stacking model
y_pred_stacking = stacking_model.predict(X_test)
if hasattr(stacking_model, "predict_proba"):
    y_pred_proba_stacking = stacking_model.predict_proba(X_test)[:, 1]
else:
    y_pred_proba_stacking = None


results['Stacking'] = {
    'Accuracy': accuracy_score(y_test, y_pred_stacking),
    'Precision': precision_score(y_test, y_pred_stacking, pos_label='Yes'),
    'Recall': recall_score(y_test, y_pred_stacking, pos_label='Yes'),
    'F1-score': f1_score(y_test, y_pred_stacking, pos_label='Yes'),
     'ROC AUC': roc_auc_score(y_test, y_pred_proba_stacking) if y_pred_proba_stacking is not None else 'N/A'
}

# Print the results
for name, metrics in results.items():
    print(f"--- {name} ---")
    for metric, value in metrics.items():
        print(f"{metric}: {value:.4f}")
    print("-" * 20)
```
